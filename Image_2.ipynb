```python
import cv2
import matplotlib.pyplot as plt

# Corrected path to an existing image file
img = cv2.imread("project_subset/New folder (2)/Train/NORMAL/IM-0125-0001.jpeg", 0)

if img is None:
    print("Error: Image not found or could not be loaded. Please check the path.")
else:
    plt.imshow(img, cmap='gray')
    plt.title("Original Image")
    plt.axis("off")
    plt.show()
```

```python
sobelx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
sobely = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)
sobel = cv2.magnitude(sobelx, sobely)

plt.imshow(sobel, cmap='gray')
plt.title("Sobel Edge Detection")
plt.axis("off")
plt.show()
```

```python
canny = cv2.Canny(img, 100, 200)
plt.imshow(canny, cmap='gray')
plt.title("Canny Edge Detection")
plt.axis("off")
plt.show()
```

```python
import numpy as np

kernelx = np.array([[-1,0,1],[-1,0,1],[-1,0,1]])
kernely = np.array([[-1,-1,-1],[0,0,0],[1,1,1]])

prewittx = cv2.filter2D(img, -1, kernelx)
prewitty = cv2.filter2D(img, -1, kernely)
prewitt = prewittx + prewitty

plt.imshow(prewitt, cmap='gray')
plt.title("Prewitt Edge Detection")
plt.axis("off")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# قراءة صورة سادسة (استبدل بالاسم اللي عندك)
img6 = cv2.imread("project_subset/New folder (2)/test/NORMAL/IM-0007-0001.jpeg", 0)

if img6 is None:
    print("Error: Image not found or could not be loaded. Please check the path and filename.")
else:
    # Sobel مع ksize=3
    sobelx_small = cv2.Sobel(img6, cv2.CV_64F, 1, 0, ksize=3)
    sobely_small = cv2.Sobel(img6, cv2.CV_64F, 0, 1, ksize=3)
    sobel_small = cv2.magnitude(sobelx_small, sobely_small)

    # Sobel مع ksize=7
    sobelx_big = cv2.Sobel(img6, cv2.CV_64F, 1, 0, ksize=7)
    sobely_big = cv2.Sobel(img6, cv2.CV_64F, 0, 1, ksize=7)
    sobel_big = cv2.magnitude(sobelx_big, sobely_big)

    # ✨ Normalization عشان يبينوا بشكل ملموس
    sobel_small_norm = cv2.normalize(sobel_small, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
    sobel_big_norm = cv2.normalize(sobel_big, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

    # عرض الصور جنب بعض
    fig, axs = plt.subplots(1, 3, figsize=(18,6))

    axs[0].imshow(img6, cmap='gray')
    axs[0].set_title("Original")
    axs[0].axis("off")

    axs[1].imshow(sobel_small_norm, cmap='gray')
    axs[1].set_title("Sobel (ksize=3, normalized)")
    axs[1].axis("off")

    axs[2].imshow(sobel_big_norm, cmap='gray')
    axs[2].set_title("Sobel (ksize=7, normalized)")
    axs[2].axis("off")

    plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# قراءة صورة وحدة للتجربة
img = cv2.imread("project_subset/New folder (2)/test/NORMAL/IM-0011-0001-0001.jpeg") # عدل اسم الصورة حسب الموجود

if img is None:
    print("Error: Image not found or could not be loaded. Please check the path.")
else:
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # تطبيق Prewitt
    kernelx = np.array([[1,0,-1],[1,0,-1],[1,0,-1]])
    kernely = np.array([[1,1,1],[0,0,0],[-1,-1,-1]])
    prewittx = cv2.filter2D(gray, -1, kernelx)
    prewitty = cv2.filter2D(gray, -1, kernely)
    prewitt = cv2.magnitude(prewittx.astype(float), prewitty.astype(float))

    # تحويل للمجال 0-255
    prewitt_norm = cv2.normalize(prewitt, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

    # تطبيق threshold منخفض وعالي
    _, prewitt_low = cv2.threshold(prewitt_norm, 10, 255, cv2.THRESH_BINARY)   # تفاصيل كثيرة
    _, prewitt_high = cv2.threshold(prewitt_norm, 150, 255, cv2.THRESH_BINARY) # حواف قوية فقط

    # عرض كل صورة لوحدها
    fig, axs = plt.subplots(1, 3, figsize=(18,6))

    axs[0].imshow(gray, cmap='gray')
    axs[0].set_title("Original")
    axs[0].axis("off")

    axs[1].imshow(prewitt_low, cmap='gray')
    axs[1].set_title("Prewitt - Low Threshold (Normalization)")
    axs[1].axis("off")

    axs[2].imshow(prewitt_high, cmap='gray')
    axs[2].set_title("Prewitt - High Threshold (Normalization)")
    axs[2].axis("off")

    plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# قراءة صورة وحدة للتجربة
img = cv2.imread("project_subset/New folder (2)/test/NORMAL/IM-0011-0001-0001.jpeg")

if img is None:
    print("Error: Image not found or could not be loaded. Please check the path.")
else:
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # تطبيق Prewitt بدون Normalization وبدون convertScaleAbs
    kernelx = np.array([[1,0,-1],
                        [1,0,-1],
                        [1,0,-1]])

    kernely = np.array([[1,1,1],
                        [0,0,0],
                        [-1,-1,-1]])

    prewittx = cv2.filter2D(gray.astype(np.float32), cv2.CV_32F, kernelx)
    prewitty = cv2.filter2D(gray.astype(np.float32), cv2.CV_32F, kernely)

    # حساب مقدار الحواف
    prewitt = cv2.magnitude(prewittx, prewitty)

    # Threshold مباشرة بدون ScaleAbs
    _, prewitt_low = cv2.threshold(prewitt, 50, 255, cv2.THRESH_BINARY)
    _, prewitt_high = cv2.threshold(prewitt, 200, 255, cv2.THRESH_BINARY)

    # تحويل فقط للعرض في matplotlib
    prewitt_low = prewitt_low.astype(np.uint8)
    prewitt_high = prewitt_high.astype(np.uint8)

    # عرض الصور
    fig, axs = plt.subplots(1, 3, figsize=(18,6))

    axs[0].imshow(gray, cmap='gray')
    axs[0].set_title("Original")
    axs[0].axis("off")

    axs[1].imshow(prewitt_low, cmap='gray')
    axs[1].set_title("Prewitt - Low Threshold")
    axs[1].axis("off")

    axs[2].imshow(prewitt_high, cmap='gray')
    axs[2].set_title("Prewitt - High Threshold")
    axs[2].axis("off")

    plt.show()
```

```python
import cv2
import numpy as np
import pandas as pd
from skimage.metrics import peak_signal_noise_ratio as psnr
from skimage.metrics import structural_similarity as ssim

# تحميل صورة وحدة للتجربة
img = cv2.imread("project_subset/New folder (2)/test/NORMAL/IM-0011-0001-0001.jpeg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Sobel
sobelx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
sobely = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
sobel = cv2.magnitude(sobelx, sobely)

# Prewitt
kernelx = np.array([[1,0,-1],[1,0,-1],[1,0,-1]])
kernely = np.array([[1,1,1],[0,0,0],[-1,-1,-1]])
prewittx = cv2.filter2D(gray, -1, kernelx)
prewitty = cv2.filter2D(gray, -1, kernely)
prewitt = cv2.magnitude(prewittx.astype(float), prewitty.astype(float))

# Canny
canny = cv2.Canny(gray, 100, 200)

# حساب PSNR و SSIM
results = {
    "Filter": ["Original", "Sobel", "Prewitt", "Canny"],
    "PSNR": [
        psnr(gray, gray),          # مقارنة الأصل مع نفسه = أعلى قيمة ممكنة
        psnr(gray, sobel),
        psnr(gray, prewitt),
        psnr(gray, canny)
    ],
    "SSIM": [
        ssim(gray, gray, data_range=gray.max()-gray.min()),  # الأصل مع نفسه = 1.0
        ssim(gray, sobel, data_range=sobel.max()-sobel.min()),
        ssim(gray, prewitt, data_range=prewitt.max()-prewitt.min()),
        ssim(gray, canny, data_range=canny.max()-canny.min())
    ]
}

# تحويل النتائج لجدول
df = pd.DataFrame(results)
print(df)
```
